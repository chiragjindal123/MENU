```python
# 1. INSTALL YOLO (run once)
!pip install ultralytics -q

# 2. SETUP DATASET (NO DRIVE, NO COPY)
import os

# 👉 Set this to your actual dataset path
# Example: if extracted in /root/hirenet/dataset
local_path = '/root/hirenet/dataset/mixed_sign_classid_handwritten_dataset_yolo_dataset'

# Check if dataset exists
if not os.path.exists(local_path):
    raise FileNotFoundError(f"❌ Dataset not found at {local_path}")
else:
    print(f"✅ Dataset found at {local_path}")

# 3. CREATE CONFIG FILE (data.yaml)
yaml_content = f"""
path: {local_path}
train: train/images
val: val/images

nc: 13
names: ['marked_circle', 'marked_number_1', 'marked_number_2', 'marked_number_3', 'marked_number_4', 'marked_number_5', 'marked_number_6', 'marked_number_7', 'marked_number_8', 'marked_number_9', 'marked_tick', 'marked_x', 'marked_hollow']
"""

# Save YAML in current directory
with open('data.yaml', 'w') as f:
    f.write(yaml_content)

print("✅ data.yaml created successfully!")
```

```python
import shutil
from pathlib import Path

try:
    from ultralytics import YOLO
except Exception as e:
    raise SystemExit("ultralytics is required. Install with: pip install ultralytics") from e

# --- MULTI-CLASS MARK DETECTION TRAINING ---
# Now trains to detect different mark types and numbers for quantity tracking

WEIGHTS = "yolov8x.pt"         # default pretrained weights
DATA_YAML = "data.yaml"        # MUST have nc: 14 and all class names
EPOCHS = 300                   # Increased epochs for multi-class learning
IMGSZ = 1280
BATCH = 8
workers = 2
RUN_NAME = "handwritten_menu_model_multiclass"
SAVE_PATH = "handwritten_mixed_menu_classid_model_best.pt"

# Advanced training parameters for better multi-class detection
PATIENCE = 50                  # Early stopping patience
OPTIMIZER = 'AdamW'            # Optimizer (SGD, Adam, AdamW)
LR0 = 0.01                     # Initial learning rate
LRF = 0.01                     # Final learning rate (lr0 * lrf)
WARMUP_EPOCHS = 3              # Warmup epochs
WEIGHT_DECAY = 0.0005          # Weight decay

def train(weights=WEIGHTS,
          data=DATA_YAML,
          epochs=EPOCHS,
          imgsz=IMGSZ,
          batch=BATCH,
          name=RUN_NAME,
          save_path=SAVE_PATH,
          patience=PATIENCE,
          optimizer=OPTIMIZER,
          lr0=LR0,
          lrf=LRF,
          warmup_epochs=WARMUP_EPOCHS,
          weight_decay=WEIGHT_DECAY):

    print("="*70)
    print("🚀 YOLO Multi-Class Mark Detection Training")
    print("="*70)
    print(f"Training Configuration:")
    print(f"  - Weights: {weights}")
    print(f"  - Data YAML: {data}")
    print(f"  - Epochs: {epochs}")
    print(f"  - Image Size: {imgsz}")
    print(f"  - Batch Size: {batch}")
    print(f"  - Optimizer: {optimizer}")
    print(f"  - Learning Rate: {lr0} -> {lr0 * lrf}")
    print(f"  - Patience: {patience}")
    print("="*70)

    # Verify data.yaml exists
    data_path = Path(data)
    if not data_path.exists():
        raise FileNotFoundError(f"Data YAML not found: {data}\n"
                              "Please ensure data.yaml has nc: 14 and all class names!")

    # Load model
    model = YOLO(weights)

    # Train with advanced parameters
    results = model.train(
        data=data,
        epochs=epochs,
        imgsz=imgsz,
        batch=batch,
        name=name,
        patience=patience,
        optimizer=optimizer,
        lr0=lr0,
        lrf=lrf,
        warmup_epochs=warmup_epochs,
        weight_decay=weight_decay,
        project="runs/detect",
        exist_ok=True,
        pretrained=True,
        verbose=True,
        # Data augmentation settings for better generalization
        hsv_h=0.015,           # HSV-Hue augmentation
        hsv_s=0.7,             # HSV-Saturation augmentation
        hsv_v=0.4,             # HSV-Value augmentation
        degrees=10.0,          # Image rotation (+/- deg)
        translate=0.1,         # Image translation (+/- fraction)
        scale=0.5,             # Image scale (+/- gain)
        shear=0.0,             # Image shear (+/- deg)
        perspective=0.0,       # Image perspective (+/- fraction)
        flipud=0.0,            # Image flip up-down (probability)
        fliplr=0.5,            # Image flip left-right (probability)
        mosaic=1.0,            # Mosaic augmentation (probability)
        mixup=0.1,             # Mixup augmentation (probability)
    )

    print("\n" + "="*70)
    print("✅ Training Complete!")
    print("="*70)

    # Copy best model to workspace root
    best_path = Path("runs") / "detect" / name / "weights" / "best.pt"
    if best_path.exists():
        dest = Path(save_path)
        dest.parent.mkdir(parents=True, exist_ok=True)
        shutil.copy(best_path, dest)
        print(f"✅ Best model copied to: {dest.resolve()}")

        # Display training results summary
        print("\n" + "="*70)
        print("📊 Training Results:")
        print("="*70)
        results_csv = Path("runs") / "detect" / name / "results.csv"
        if results_csv.exists():
            import pandas as pd
            try:
                df = pd.read_csv(results_csv)
                if not df.empty:
                    last_row = df.iloc[-1]
                    print(f"Final Metrics (Epoch {int(last_row.get('epoch', epochs))}):")

                    # Show key metrics
                    metrics_to_show = [
                        ('metrics/precision(B)', 'Precision'),
                        ('metrics/recall(B)', 'Recall'),
                        ('metrics/mAP50(B)', 'mAP@50'),
                        ('metrics/mAP50-95(B)', 'mAP@50-95'),
                    ]

                    for col, label in metrics_to_show:
                        if col in df.columns:
                            value = last_row[col]
                            print(f"  {label}: {value:.4f}")
            except Exception as e:
                print(f"Could not read results.csv: {e}")

        print("="*70)
        print(f"\n💡 Next Steps:")
        print(f"   1. Test the model: python test_model.py")
        print(f"   2. Run live detection: python live.py")
        print(f"   3. Check detailed results in: runs/detect/{name}/")
        print("="*70)
    else:
        print(f"❌ Could not find {best_path}. Check runs/detect/{name}/")
        print(f"   Training may have failed or path has changed.")

if __name__ == "__main__":
    train()
```
